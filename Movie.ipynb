```python
movies.drop_duplicates(inplace= True)
```

```python
movies.dropna(subset = ['genres'], inplace = True)
```

```python
movies['profit'] = movies['revenue'] - movies['budget']
```

```python
#drop unecassary columns
```

```python
movie_genres = movies[['id', 'popularity', 'budget', 'original_title', 'runtime', 'genres', 'release_year', 'profit', 'vote_count', 'vote_average', 'revenue', 'genres_split']]
```

```python
movie_genres.head()
```

```python
from pandas import Series, DataFrame

```

```python
#split genre column to improve workability
```

```python
split = movie_genres['genres'].str.split('|').apply(Series,1).stack()
split.index = split.index.droplevel(-1)
split.name = 'genres_split'
del movie_genres['genres']
movie_genres = movie_genres.join(split)
```

```python
movie_genres
```

```python
movies_avg = movie_genres[['budget', 'revenue', 'genres_split', 'profit','popularity']].groupby('genres_split').mean()
pd.options.display.float_format = '{:2f}'.format
```

```python
movies.drop_duplicates(inplace= True)
movies['profit'] = movies['revenue'] - movies['budget']
movie_genres = movies[['id', 'popularity', 'budget', 'original_title', 'runtime', 'genres', 'release_year', 'profit', 'vote_count', 'vote_average', 'revenue']]
```

```python
movie_genres.head()
```

```python
#1.The best movies according to vote avg. return high profit and revenue
```

```python
movie_count = movie_genres[movie_genres['vote_count'] >= 50]

movie_count.corr(method = 'spearman', numeric_only = True)
```

```python
import seaborn as sns

sns.regplot(x = 'vote_average', y = 'profit', data = movie_count, line_kws = {"color": 'red'})
```

```python
sns.regplot(x = 'vote_average', y = 'revenue', data = movie_count, line_kws = {"color": 'red'})
```

```python
#2.The best movies according to popularity return high profit and revenue
```

```python
movie_count.corr(method = 'spearman', numeric_only = True)
```

```python
#profit per genre per year
```

```python
time_genre = pd.DataFrame(movie_genres.groupby(['release_year', 'genres_split'])['profit'].mean())
```

```python
final_genre = pd.pivot_table(time_genre, values = 'profit', index = ['genres_split'], columns = ['release_year'])
```

```python
sns.set(rc = {'figure.figsize': (15,10)})
sns.heatmap(final_genre, cmap = 'YlGnBu', linewidths = .5)
plt.title("Genres by profit per year")
```
